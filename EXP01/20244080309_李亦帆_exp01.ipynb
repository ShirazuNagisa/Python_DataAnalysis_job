```python
print("8 / 3  =", 8 / 3, type(8 / 3).__name__)
print("8 // 3 =", 8 // 3, type(8 // 3).__name__)
print("8 % 3  =", 8 % 3, type(8 % 3).__name__)
print("2 ** 7 =", 2 ** 7, type(2 ** 7).__name__)
```

## 实验2 函数、条件、循环与递归

```python
def score_level(score):
    if score < 0 or score > 100:
        raise ValueError(f"分数必须在 0~100 之间：{score}")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"


for score in (100, 90, 89, 80, 79, 60, 59, 0):
    print(score, score_level(score))
for bad in (-1, 101):
    try:
        score_level(bad)
    except ValueError as error:
        print(bad, "->", error)
```

```python
total_for = 0
for number in range(1, 101):
    total_for += number

total_while, number = 0, 1
while number <= 100:
    total_while += number
    number += 1

print(total_for, total_while, total_for == total_while)
```

```python
def factorial_recursive(n):
    if n < 0:
        raise ValueError(n)
    return 1 if n <= 1 else n * factorial_recursive(n - 1)


def factorial_iterative(n):
    if n < 0:
        raise ValueError(n)
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result


for n in (0, 1, 5, 10):
    print(n, factorial_recursive(n), factorial_iterative(n))
```

```python
def summary(values):
    if not values:
        raise ValueError("values 不能为空")
    return min(values), max(values), sum(values) / len(values)


lowest, highest, average = summary([85, 58, 92, 76, 60, 89])
print(lowest, highest, round(average, 3), summary([1, 2, 3, 4]))
```

```python
def sum_all(*args):
    return sum(args)


print(sum_all(), sum_all(5), sum_all(1, 2, 3, 4.5), sum_all(*[10, 20, 30]))
```

## 实验4 文件读写、异常处理与模块

```python
import os

print(os.getcwd())
data_dir = os.path.join("exp01_data")
os.makedirs(data_dir, exist_ok=True)
data_path = os.path.join(data_dir, "students.txt")
report_path = os.path.join(data_dir, "report.txt")
pickle_path = os.path.join(data_dir, "students.pkl")
```

```python
student_text = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete"""

with open(data_path, "w", encoding="utf-8") as file:
    file.write(student_text + "\n")

with open(data_path, encoding="utf-8") as file:
    lines = [line.strip() for line in file if line.strip()]

print(lines[0])
for line in lines[1:]:
    print(line, line.split(","))
```

```python
def parse_student(line):
    fields = line.split(",")
    if len(fields) != 2:
        raise ValueError(f"字段数量应为 2，实际 {len(fields)}")
    name, score_text = fields[0].strip(), fields[1].strip()
    if not name:
        raise ValueError("姓名为空")
    try:
        score = int(score_text)
    except ValueError:
        raise ValueError(f"成绩不是整数：{score_text!r}")
    if not 0 <= score <= 100:
        raise ValueError(f"成绩超出 0~100：{score}")
    return {"name": name, "score": score}


valid_records, invalid_records = [], []
with open(data_path, encoding="utf-8") as file:
    next(file)                                  # 跳过表头
    for line_number, line in enumerate(file, start=2):
        line = line.strip()
        if not line:
            continue
        try:
            valid_records.append(parse_student(line))
        except ValueError as error:
            invalid_records.append((line_number, line, str(error)))
            continue

print(len(valid_records), valid_records)
for item in invalid_records:
    print(item)
```

```python
average = sum(record["score"] for record in valid_records) / len(valid_records)
pass_rate = sum(1 for record in valid_records if record["score"] >= 60) / len(valid_records)
with open(report_path, "w", encoding="utf-8") as file:
    file.write(f"有效记录数: {len(valid_records)}\n")
    file.write(f"无效记录数: {len(invalid_records)}\n")
    file.write(f"平均分: {average:.2f}\n")
    file.write(f"及格率: {pass_rate:.2%}\n")

print(open(report_path, encoding="utf-8").read())
```

```python
demo_path = os.path.join(data_dir, "mode_demo.txt")
with open(demo_path, "w", encoding="utf-8") as file:
    file.write("w 模式写入第一次\n")
with open(demo_path, "w", encoding="utf-8") as file:
    file.write("w 模式写入第二次\n")
with open(demo_path, "a", encoding="utf-8") as file:
    file.write("a 模式追加第一次\n")
with open(demo_path, "a", encoding="utf-8") as file:
    file.write("a 模式追加第二次\n")

print(open(demo_path, encoding="utf-8").read())
```

`w` 打开即清空原内容、从头写，所以第二次 `w` 覆盖了第一次；`a` 保留原内容、在末尾追加，两次追加都保留。演示写在 `mode_demo.txt`，不影响 `report.txt`。

## 实验5 类、继承与对象复制

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"{self.name}({self.score}) {'及格' if self.is_passed() else '不及格'}"


students = [Student(record["name"], record["score"]) for record in valid_records]
for student in students:
    print(student, student.is_passed())
```

```python
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return f"{super().__str__()} 研究方向: {self.research_topic}"


graduate = GraduateStudent("Alice", 85, "数据分析")
print(graduate, isinstance(graduate, Student))
```

```python
import copy

original = {"name": "Alice", "scores": [85, 90]}
assigned = original
assigned["scores"].append(95)
print("直接赋值:", original, assigned)

original = {"name": "Alice", "scores": [85, 90]}
shallow = copy.copy(original)
shallow["scores"].append(95)
print("copy():", original, shallow, "内层共享:", shallow["scores"] is original["scores"])

original = {"name": "Alice", "scores": [85, 90]}
deep = copy.deepcopy(original)
deep["scores"].append(95)
print("deepcopy():", original, deep, "内层共享:", deep["scores"] is original["scores"])
```

直接赋值是同一个对象，改内层列表两边都变；`copy.copy()` 只新建外层，内层列表仍共享（内层共享 `True`）；`copy.deepcopy()` 递归复制，内层也独立（内层共享 `False`）。三段比较都从新建的 `original` 开始。
